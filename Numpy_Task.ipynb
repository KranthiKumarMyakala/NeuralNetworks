```python
import pandas as pd
import numpy as np

weights1 = [[2, 8, 5, 1],
              [5, 9, 6, 5],
              [2, 2, 1, 7],
              [1, 2, 3, 4]] 
inputs = [
    1, 2, 3, 2
]
neuron_bias = [2 , 3, 0,7] 

weights2 = [[2, 8, 5, 1],
              [5, 9, 6, 5],
              [2, 2, 1, 7],
              [1, 2, 3, 4]] 

neuron_bias1 = [2 , 3, 0,4]

layer_outputs = np.dot(inputs, np.array(weights1).T) + neuron_bias
#relu activation function
layer_outputs = np.maximum(0, layer_outputs)
print("otutput of layer 1 : ", *layer_outputs, sep = " ")

layer_outputs1 = np.dot(layer_outputs, np.array(weights2).T) + neuron_bias1
#relu activation function
layer_outputs1 = np.maximum(0, layer_outputs1)
print("output of layer 2 : ", *layer_outputs1, sep = " ")




```
