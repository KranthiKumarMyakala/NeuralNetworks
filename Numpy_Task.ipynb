```python
import pandas as pd
import numpy as np

# Create a single neuron with 3 inputs and a bias
weights = [[0.2, 0.8, -0.5, 1.0],
              [0.5, -0.91, 0.26, -0.5],
              [-0.26, -0.27, 0.17, 0.87],
              [0.1, -0.14, 0.5, -0.12],
              [0.5, 0.1, -0.5, -0.2]] # 5 neurons, 4 inputs each
inputs = [
    [1.0, 2.0, 3.0, 2.5],
    [2.0, 5.0, -1.0, 2.0],
    [-1.5, 2.7, 3.3, -0.8],
    [1.0, 2.0, 3.0, 2.5],
    [2.0, 5.0, -1.0, 2.0],
    [-1.5, 2.7, 3.3, -0.8] # 6 samples, 4 features each
]
neuron_bias = [2.0 , 3.0, 0.5, 1.0, -1.0] # 5 neurons, 1 bias each

output = np.dot(inputs, np.array(weights).T) + neuron_bias # 6 samples, 5 neurons
print(output)
```
